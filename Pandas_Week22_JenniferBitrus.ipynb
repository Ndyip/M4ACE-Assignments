```python
#Countries in Africa
africa = df[df["Continent"] == "Africa"]
africa.head(10)
```

```python
#Countries with population > 100M in 2022
large_population = df[df["2022 Population"] > 100000000]
large_population[["Country", "2022 Population"]]
```

```python
#Countries with growth rate above 1.02
high_growth = df[df["Growth Rate"] > 1.02]
high_growth[["Country", "Growth Rate"]]
```

```python
#Number of countries per continent
countries_per_continent = df.groupby("Continent")["Country"].count()
countries_per_continent
```

```python
#Average population by continent 2022
avg_population = df.groupby("Continent")["2022 Population"].mean()
avg_population
```

```python
#Top 5 countries 
top5 = df.sort_values(by="2022 Population", ascending=False)
top5[["Country", "2022 Population"]].head()
```
